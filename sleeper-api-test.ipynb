```python
import requests

API = "https://api.sleeper.app/v1"
get = lambda p: requests.get(f"{API}{p}", timeout=30).json()

state = get("/state/nfl")
user = get("/user/koenifra")
leagues = get(f"/user/{user['user_id']}/leagues/nfl/{state['season']}")
league_id = leagues[0]["league_id"]
matchups = get(f"/league/{league_id}/matchups/{state['week']}")
print(matchups[0].keys())  # inspect players_points, starters, etc.
```
